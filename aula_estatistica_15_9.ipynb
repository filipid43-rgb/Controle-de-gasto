pascal

```python
from scipy.stats import nbinom

r =3
p = 0.05
k = 50

x = k -r #número de fracassos

prob = nbinom.pmf(x, r, p)

print(f"Probabilidade da minha 3ª venda acontecer na 50ª ligação:")
print(f"{prob*100:.4f}%")
```

hypergeom

```python
from scipy.stats import hypergeom

N = 1000
K = 50
n = 50
k = 3

prob = hypergeom.pmf(k, N, K,n)

print(f"Probabilidade de exatamente {K} peças defeituosas na amostra:")
print(f"{prob * 100:.4f}%")
```

poisson

```python
from scipy.stats import poisson

lambda_param = 100
k = 130

prob = poisson.pmf(k, lambda_param)

print(f"Probabilidade de {k} carros passarem é: ")
print(f"{prob * 100:.10f}%")
```

```python
#parametros da distribuiçao uniforme
a = 485
b = 515

#cálculo
f_x = 1 / (b -a)

#resultado
print(f"f(x) = 1 / ({b} - {a}) = {f_x:.6f}")
print(f"Em porcetagem: {f_x * 100:.2f}%")
```
